```python
#Se carga una base de datos con registro de criminalidad
df = pd.read_csv('MPS_Borough_Level_Crime_Historic.csv', header=0, encoding='utf-8')
df.head(3)
```

```python
df.major_category.unique()
```

```python
#Filtro para incluir solo datos de 'Violence Against the Person'
filtered = df['major_category'] == 'Violence Against the Person'
violence = df[filtered]
violence
```

```python
#Aquí agrupamos la información por barrio:
df_new = violence.groupby('borough').sum()
df_new.head()
```

```python
data
```

```python
merged = data.set_index('NAME').join(df_new)
merged = merged.reindex(merged.index.rename('borough'))
```

```python
merged.head(2)
```

```python
plot = merged.plot(column='200804', cmap='OrRd')
fig = plot.get_figure()
```

```python
plot = merged.plot(column='201601', cmap='OrRd')
fig = plot.get_figure()
```

```python
#Crear un gif que contenga multiples mapas(uno por fecha)

output_path = 'maps'

#contador para el bucle
i=0

#lista de años
list_of_years =['200807','200907','201007','201107','201207','201307','201407','201507','201607']

#rango de valores para el mapa
vmin, vmax = 200, 1200

#bucle para generar los mapas
for year in list_of_years:

    #crear un mapa para cada fecha indicada
    fig = merged.plot(column=year, cmap='Purples',figsize=(10,10), linewidth=0.8, edgecolor='0.8', vmin=vmin, vmax=vmax,
                      legend=True, norm=plt.Normalize(vmin=vmin, vmax=vmax))

    # remover los valores en los ejes
    fig.axis('off')

    fig.set_title('Violent crimes in London', \
                  fontdict={'fontsize':'25',
                            'fontweight': '3'})
    #crear una nota para indicar la fecha
    only_year = year[:4]

    #posición de la nota

    fig.annotate(only_year,
                 xy=(0.1, .225), xycoords='figure fraction',
                 horizontalalignment='left', verticalalignment='top',
                 fontsize=35)

    #Para guardar las figuras como png
    filepath = os.path.join(output_path, only_year+'violence.png')
    chart = fig.get_figure()
    chart.savefig(filepath, dpi=100)
```

```python
import imageio.v2 as imageio
import os
import numpy as np

folder = 'maps'
files = [f'{folder}/{file}' for file in sorted(os.listdir(folder))]

# Load images explicitly into a list of NumPy arrays
# Ensure each image is a NumPy array
images = [np.array(imageio.imread(file)) for file in files]

imageio.mimwrite('movie.gif', images, fps=1)
```
